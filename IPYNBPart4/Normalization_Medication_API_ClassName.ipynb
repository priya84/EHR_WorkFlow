```python
D_Epilepsy_Drug_List1 = spark.read.parquet("file:/home/o_suchsi/work/Oklahoma State/Priya/epilepsy/Finalized_Normalized__Distinct_DrugNames_ClassNames")
```

```python
combined_df.printSchema()
D_Epilepsy_Drug_List1.printSchema()
```

```python
DD_combined_df = combined_df.select("drugname").distinct()
DD_Epilepsy_Drug_List1 = D_Epilepsy_Drug_List1.select("drugname").distinct()
```

```python
# Import required functions
from pyspark.sql.functions import col

# Filter common records based on drugname column
common_drugs_compare = DD_combined_df.join(DD_Epilepsy_Drug_List1, "drugname", "inner")

# Display the count of distinct drugname
distinct_drugname_count = common_drugs_compare.select("drugname").distinct().count()
print("Count of distinct drugname in common records:", distinct_drugname_count)
```

```python
# Perform a left anti join to find records in DD_combined_df not present in DD_Epilepsy_Drug_List1
missing_drugnames_df = DD_combined_df.join(DD_Epilepsy_Drug_List1, "drugname", "left_anti")

# Count the number of such records
missing_drugname_count = missing_drugnames_df.count()
print("Count of records with drugname values present in DD_combined_df but not in DD_Epilepsy_Drug_List1:", missing_drugname_count)
```

```python
# Perform a left anti join to find records in DD_Epilepsy_Drug_List1 not present in DD_combined_df
missing_drugnames_df1 = DD_Epilepsy_Drug_List1.join(DD_combined_df, "drugname", "left_anti")

# Count the number of such records
missing_drugname_count1 = missing_drugnames_df1.count()
print("Count of records with drugname values present in DD_Epilepsy_Drug_List1 but not in DD_combined_df:", missing_drugname_count1)
```

```python
# Assuming df is your DataFrame and cols_to_drop is a list of column names you want to drop
cols_to_drop = ["ClassId_Hierarchy", "Length", "level1", "level2", "level3", "level4", "DrugList_Hierarchy", "DrugList_M1"]

# Drop the specified columns
Total_Drug_List1 = D_Epilepsy_Drug_List1.drop(*cols_to_drop)

# Show the DataFrame after dropping columns
Total_Drug_List1.show(truncate=False)
Total_Drug_List1.printSchema()
print("Count of Total_Drug_List1", Total_Drug_List1.count())
```

```python
Total_Drug_List1 = Total_Drug_List1.withColumnRenamed("ClassName_Hierarchy", "classname")
Total_Drug_List1.printSchema()
```

```python
from pyspark.sql.functions import when
# Perform left join to match drugnames and extract corresponding classname
Cohort_Med_df = Epilepsy_Cohort_Paired_New.join(Total_Drug_List1, Epilepsy_Cohort_Paired_New.drugname == Total_Drug_List1.drugname, "left_outer") \
    .select(Epilepsy_Cohort_Paired_New.personid, Epilepsy_Cohort_Paired_New.drugname, Total_Drug_List1.classname) \
    .orderBy(Epilepsy_Cohort_Paired_New.personid)  # Ensure the order is maintained as in Epilepsy_Cohort_Paired_New

# Mark classname as Null for unmatched drugnames
Cohort_Med_df = Cohort_Med_df.withColumn("classname", when(Cohort_Med_df.classname.isNull(), "Null").otherwise(Cohort_Med_df.classname))

# Show the final resulted dataframe
Cohort_Med_df.show(truncate=False)
print("Count of Cohort_Med_df", Cohort_Med_df.count())
```

```python
Cohort_Med_df.write.mode("overwrite").parquet("file:/home/o_suchsi/work/Oklahoma State/Priya/epilepsy/Epilepsy_Cohort_Med_df_stack")
```

```python
# print(Cohort_Med_df.select("drugname").distinct().count())
# Count non-null drugnames
non_null_count = Cohort_Med_df.filter(Cohort_Med_df.drugname.isNotNull()).select("drugname").distinct().count()

print("Count of non-null drugnames:", non_null_count)
```

```python
from pyspark.sql.functions import when
# Perform left join to match drugnames and extract corresponding classname
Control_Med_df = Epilepsy_Control_Paired_New.join(Total_Drug_List1, Epilepsy_Control_Paired_New.drugname == Total_Drug_List1.drugname, "left_outer") \
    .select(Epilepsy_Control_Paired_New.personid, Epilepsy_Control_Paired_New.drugname, Total_Drug_List1.classname) \
    .orderBy(Epilepsy_Control_Paired_New.personid)  # Ensure the order is maintained as in Epilepsy_Cohort_Paired_New

# Mark classname as Null for unmatched drugnames
Control_Med_df = Control_Med_df.withColumn("classname", when(Control_Med_df.classname.isNull(), "Null").otherwise(Control_Med_df.classname))

# Show the final resulted dataframe
Control_Med_df.show(truncate=False)
print("Count of Control_Med_df", Control_Med_df.count())
```

```python
Control_Med_df.write.mode("overwrite").parquet("file:/home/o_suchsi/work/Oklahoma State/Priya/epilepsy/Epilepsy_Control_Med_df_stack")
```

```python
# print(Cohort_Med_df.select("drugname").distinct().count())
# Count non-null drugnames
non_null_count_Ctl = Control_Med_df.filter(Control_Med_df.drugname.isNotNull()).select("drugname").distinct().count()

print("Count of non-null control drugnames:", non_null_count_Ctl)
```
