```python
import solara
from pathlib import Path
import io
import zipfile

prefix = Path("/home/o_suchsi/work/Oklahoma State/Priya/epilepsy/FilePart1")

# Build an in-memory zip of EVERYTHING under prefix (files + folders)
zip_buffer = io.BytesIO()
with zipfile.ZipFile(zip_buffer, mode="w", compression=zipfile.ZIP_DEFLATED) as zf:
    for p in prefix.rglob("*"):
        # Preserve folder structure inside the zip
        arcname = p.relative_to(prefix)
        if p.is_dir():
            # Add folder entry (optional but helps some unzip tools show empty dirs)
            zf.writestr(str(arcname) + "/", "")
        else:
            zf.write(p, arcname=str(arcname))

zip_buffer.seek(0)
zip_data = zip_buffer.getvalue()

display(solara.FileDownload(
    data=zip_data,
    label=f"Download {prefix.name}.zip",
    filename=f"{prefix.name}.zip"
))
```

```python
import solara
import shutil
import os

def zip_folder(folder_path, output_filename):
    """Creates a zip archive of the given folder."""
    shutil.make_archive(output_filename, 'zip', folder_path)
    return output_filename + ".zip"

# Define the root folder to download
# root = "checkpoint9"
root = "priya-smallset-finalF1_Numbered_NoNullStr.parquet"
zip_filename = "priya-smallset-finalF1_Numbered_NoNullStr"  # Name of the zip file (without extension)

# Ensure the folder exists before proceeding
if not os.path.exists(root):
    raise FileNotFoundError(f"Folder '{root}' does not exist.")

# Create a ZIP archive of the folder
zip_path = zip_folder(root, zip_filename)

# Read the ZIP file as binary
with open(zip_path, 'rb') as f:
    zip_data = f.read()

# Display the zip file as a downloadable file
display(solara.FileDownload(data=zip_data, label=f"Download {zip_filename}.zip", filename=f"{zip_filename}.zip"))

# Cleanup: Remove the generated zip file after serving
os.remove(zip_path)
```
