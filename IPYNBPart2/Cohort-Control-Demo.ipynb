```python
import numpy as np
import scipy.stats as st
from pyspark.sql import functions as F

def generate_age_data(cohort_demo, age_counts):
    # Initialize empty lists to store age values and weights
    age_values = []
    weights = []

    # Define age ranges and their lower/upper bounds
    age_ranges = ['age<18', '18<=age<44', '44<=age<60', 'age>=60']
    age_bounds = [(0, 18), (18, 44), (44, 60), (60, 100)]  # Use a large upper bound for age>=60

    # Extract age values and weights from age ranges and counts
    for age_range, count in zip(age_ranges, age_counts):
        # Find the matching age bounds for the range
        lower_bound, upper_bound = age_bounds[age_ranges.index(age_range)]

        # Extract age values from the "age" column of cohort_demo
        age_column = cohort_demo.filter((F.col("age") >= lower_bound) & (F.col("age") < upper_bound)).select("age")

        # Convert age values to a list
        age_list = [row.age for row in age_column.collect()]

        # Extend the age_values list with age values based on the count
        age_values.extend(age_list * count)

        # Assign weights (counts) to each age value
        weights.extend([count] * (len(age_list) * count))

    return age_values, weights

def cohort_distribution(spark, cohort_demo, age_counts):
    # Generate age data
    age_values, weights = generate_age_data(age_counts)
    
    # Create a DataFrame to store age data
    age_data = spark.createDataFrame(list(zip(age_values, weights)), ["generated_age", "generated_weight"])
    
    # Generate a unique identifier (personid) for each row in age_data
    age_data = age_data.withColumn("personid", F.monotonically_increasing_id())
    
    # Join age data with cohort_demo
    cohort_demo = cohort_demo.join(age_data, on="personid", how="inner")
    
    # Check if cohort_demo has rows after the join
    if cohort_demo.count() > 0:
        # Calculate mean and standard error for age
        mean_age = cohort_demo.groupBy().agg(F.avg("generated_age")).collect()[0][0]
        
        # Calculate the standard error using a weighted formula
        std_error_age = np.sqrt(cohort_demo.select(F.sum(F.pow(F.col("generated_age") - mean_age, 2) * F.col("generated_weight"))).collect()[0][0] / cohort_demo.select(F.sum("generated_weight")).collect()[0][0])
        
        # Calculate the confidence interval (assuming normal distribution)
        z_score = st.norm.ppf(0.90)  # 90% confidence interval
        margin_of_error = z_score * std_error_age
        lower_bound = mean_age - margin_of_error
        upper_bound = mean_age + margin_of_error

        # Display the results
        print(f"Mean Age: {mean_age:.2f}")
        print(f"90% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
    else:
        print("No matching data found in the cohort_demo DataFrame.")

# Define the age range counts
age_counts = [27733, 53674, 30728, 78410]

# Call the cohort_distribution function with the age counts
cohort_distribution(spark, Epilepsy_Cohort_Demo_R, age_counts)
```
