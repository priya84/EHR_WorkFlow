```python
TBI_EPI.createOrReplaceTempView('TBI_EPI')
EPI = spark.sql("""
select * from TBI_EPI
where conditioncode like 'G40%' or
conditioncode like '345%'
""")
```

```python
TBI = spark.sql("""
select * from TBI_EPI
where conditioncode in ('Z87.820','S02.1','R56.1') or
conditioncode like 'S06%' or
conditioncode like 'S07%' or
conditioncode like 'S08%' or
conditioncode like 'S09%' or
conditioncode like 'G44.3%' or
conditioncode like '854.%' or
conditioncode like '851.%' or
conditioncode like '852.%' or
conditioncode like '853.%'
""")
```

```python
EPI.cache()
EPI.count()
```

```python
EPI_personid = EPI.select('personid').distinct()
TBI_EPI_personid = TBI_EPI.select('personid').distinct()
EPI_med_personid = EPI_med.select('personid').distinct()
```

```python
EPI_total = EPI_personid.unionByName(EPI_med_personid).distinct()
TBI_personid = TBI.select('personid').distinct()
TBI_with_EPI = TBI_personid.join(EPI_total,'personid').distinct()
```

```python
print('The number of TBI patients is: ',TBI_personid.count())
print('The number of TBI patients with EPI is: ',TBI_with_EPI.count())
print('The number of EPI patients is: ',EPI_total.count())
```
