```python
# Query to join the tables
sample_query = """
SELECT DISTINCT
    lab1.value, lab1.modifier, lab2.refLowtype, lab2.refLowRange, lab3.refHightype, lab3.refHighRange,
    lab4.personid, lab4.labcode, lab4.interpretation, lab4.servicedate, lab5.refLowtextvalue, lab5.refHightextvalue
FROM
    lab1
JOIN
    lab4 ON lab1.personid = lab4.personid AND lab1.labcode = lab4.labcode AND lab1.servicedate = lab4.servicedate
JOIN
    lab2 ON lab1.personid = lab2.personid AND lab1.labcode = lab2.labcode AND lab1.servicedate = lab2.servicedate
JOIN
    lab3 ON lab1.personid = lab3.personid AND lab1.labcode = lab3.labcode AND lab1.servicedate = lab3.servicedate
JOIN
    lab5 ON lab1.personid = lab5.personid AND lab1.labcode = lab5.labcode AND lab1.servicedate = lab5.servicedate
"""

# Execute the query and display the result
sample_result = spark.sql(sample_query)
sample_result.show()
```
