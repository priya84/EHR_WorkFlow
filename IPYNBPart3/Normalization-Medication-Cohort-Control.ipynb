```python
#############################Final : Convert the Non-Proprietary to ndc_dict with 1 EPC Class#####################################################
import pandas as pd
import math

# #Final List from cohort and control med combined list
df = pd.read_csv("file:///home/o_suchsi/work/Oklahoma State/Priya/epilepsy/productcleaned.csv")
# df = pd.read_csv("file:///home/o_suchsi/work/Oklahoma State/Priya/epilepsy/productcleanedlist.csv")

# Create an empty dictionary
ndc_dict = {}

# Iterate through the DataFrame rows
for index, row in df.iterrows():
    nonprop_name = row['Cleaned Items']
#     nonprop_name = row['PROPRIETARYNAME']
    pharm_classes = row['PHARM_CLASSES']
    
    # If the 'NONPROPRIETARYNAME' is not in the dictionary, add it as a key
    # and initialize the value as an empty list
    if nonprop_name not in ndc_dict:
        ndc_dict[nonprop_name] = []
    
    # Check if the 'PHARM_CLASSES' value is not NaN
    if not isinstance(pharm_classes, float):
        # Split the 'PHARM_CLASSES' value by commas and store it as a list
        pharm_classes_list = [item.strip() for item in pharm_classes.split(',')]
        
        # Extract and append only values ending with "[EPC]"
        pharm_classes_epc = [item for item in pharm_classes_list if item.endswith("[EPC]")]
        
        # If there are multiple values ending with "[EPC]", keep the one with the shorter length
        if len(pharm_classes_epc) > 1:
            shortest_length_item = min(pharm_classes_epc, key=len)
            pharm_classes_epc = [shortest_length_item]
    
        # Append the list of 'PHARM_CLASSES' values ending with "[EPC]" to the list of values for the key
        ndc_dict[nonprop_name].extend(pharm_classes_epc)

# If you want to remove duplicates in the 'PHARM_CLASSES' values, you can do so
for key, value in ndc_dict.items():
    ndc_dict[key] = list(set(value))
# with only the item having the shortest length when multiple items are found
# print(ndc_dict)
# Convert the dictionary
ndc_dict = {key: value[0] if value else None for key, value in ndc_dict.items()}
# Print the converted dictionary
print("##################################converted_dict########################################## ")
print(ndc_dict)
```
